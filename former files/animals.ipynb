# Class of animals

Immediately earn victory points according to the number of animals present on the tile. Should the player already have an animal tile of the same type in the pasture (region of connected green tiles), then he receives the number of animals on the tile + the number of animals of the same type already in the pasture.

```python
from enum import Enum

class AnimalType(Enum):
    CATTLE = "cattle"
    CHICKEN = "chicken"
    GOAT = "goat"
    PIG = "pig"
    SHEEP = "sheep"

class Animal:
    def __init__(self, type: AnimalType, quantity: int, color: Color) -> None:
        self.type = type
        self.quantity = quantity
        self.color = color
```
